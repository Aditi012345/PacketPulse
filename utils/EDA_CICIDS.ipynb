```python
def get_high_corr_pairs(corr_matrix, threshold=0.9):
    pairs = []
    cols = corr_matrix.columns
    for i in range(len(cols)):
        for j in range(i+1, len(cols)):
            val = corr_matrix.iloc[i, j]
            if pd.notna(val) and abs(val) >= threshold:
                pairs.append((cols[i], cols[j], round(val, 3)))
    return pd.DataFrame(pairs, columns=['feature_1', 'feature_2', 'correlation']).sort_values(
        'correlation', ascending=False, key=abs)

high_corr = get_high_corr_pairs(corr_matrix, threshold=0.9)
print(high_corr.to_string())
```

```python
target_corr = full_df[numerical_cols].corrwith(full_df['attack_binary'])

high_corr['corr_with_target_1'] = high_corr['feature_1'].map(target_corr)
high_corr['corr_with_target_2'] = high_corr['feature_2'].map(target_corr)
print(high_corr.to_string())
```

```python
def detect_outliers_iqr(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = df[(df[column] < lower_bound) | (df[column] > upper_bound)]
    return outliers, lower_bound, upper_bound

iqr_summary = []
for col in numerical_cols:
    outliers, lb, ub = detect_outliers_iqr(full_df, col)
    iqr_summary.append({
        'column': col,
        'lower_bound': lb,
        'upper_bound': ub,
        'num_outliers': len(outliers),
        'pct_outliers': round(len(outliers) / len(full_df) * 100, 2)
    })

iqr_df = pd.DataFrame(iqr_summary).sort_values('pct_outliers', ascending=False)
print(iqr_df.to_string())
```
