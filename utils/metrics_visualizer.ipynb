```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv("../data/cicids_model_comparison.csv")
df
```

```python
plt.figure(figsize=(10,5))
plt.bar(df['Model'], df['Test_Accuracy'])
plt.xlabel("Models")
plt.ylabel("Test Accuracy")
plt.title("Model Comparison – Test Accuracy")
plt.xticks(rotation=45)
plt.show()
```

```python
metrics = ['Test_Accuracy', 'Precision', 'Recall', 'F1_Score']

x = np.arange(len(df['Model']))
width = 0.2

plt.figure(figsize=(12, 6))

for i, metric in enumerate(metrics):
    bars = plt.bar(x + i * width, df[metric], width, label=metric,color = colors[i])

    for bar in bars:
        value = bar.get_height()
        plt.text(
            bar.get_x() + bar.get_width() / 2,
            value,
            "{:.3f}".format(value),
            ha='center',
            va='bottom',
            fontsize=9
        )

plt.xticks(x + width, df['Model'], rotation=45)
plt.ylabel("Score")
plt.title("Model Performance Comparison")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.plot(df['Model'], df['Test_Accuracy'], marker='o', label='Accuracy')
plt.plot(df['Model'], df['Precision'], marker='o', label='Precision')
plt.plot(df['Model'], df['Recall'], marker='o', label='Recall')
plt.plot(df['Model'], df['F1_Score'], marker='o', label='F1 Score')

plt.xlabel("Models")
plt.ylabel("Score")
plt.title("Model Performance Trend")
plt.xticks(rotation=45)
plt.legend()
plt.grid()
plt.show()
```

```python
metrics = ['Test_Accuracy', 'Precision', 'Recall', 'F1_Score']
N = len(metrics)

angles = np.linspace(0, 2*np.pi, N, endpoint=False).tolist()
angles += angles[:1]

plt.figure(figsize=(7,7))

for i, row in df.iterrows():
    values = row[metrics].tolist()
    values += values[:1]
    plt.polar(angles, values, marker='o', label=row['Model'])

plt.xticks(angles[:-1], metrics)
plt.title("Model Comparison Radar Chart")
plt.legend(loc='upper right', bbox_to_anchor=(1.25, 1.05))
plt.show()
```

```python
plt.figure(figsize=(12, 6))
plt.bar(df["Model"], df["Train_Accuracy"], label="Train Accuracy")
plt.bar(df["Model"], df["Test_Accuracy"], label="Test Accuracy", alpha=0.7)
plt.xticks(rotation=45)
plt.ylabel("Accuracy")
plt.title("Train vs Test Accuracy Comparison")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(14, 6))

x = range(len(df))
width = 0.25

plt.bar([p - width for p in x], df['Precision'], width=width, label='Precision')
plt.bar(x, df['Recall'], width=width, label='Recall')
plt.bar([p + width for p in x], df['F1_Score'], width=width, label='F1 Score')

plt.xticks(x, df['Model'], rotation=45)
plt.ylabel("Scores")
plt.title("Precision, Recall, F1 Score Comparison")
plt.legend()
plt.tight_layout()
plt.show()
```
