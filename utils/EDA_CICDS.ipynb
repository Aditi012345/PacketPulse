```python
# let us change the column names so it matches all the datasets
```

```python
for column in monday:
    monday.rename(columns={column:column.strip()},inplace=True)
    
for column in tuesday:
    tuesday.rename(columns={column:column.strip()},inplace=True)
    
for column in wednesday:
    wednesday.rename(columns={column:column.strip()},inplace=True)
    
for column in thursday_morning:
    thursday_morning.rename(columns={column:column.strip()},inplace=True)
    
for column in thursday_afternoon:
    thursday_afternoon.rename(columns={column:column.strip()},inplace=True)
    
for column in friday_morning:
    friday_morning.rename(columns={column:column.strip()},inplace=True)
    
for column in friday_afternoon_ddos:
    friday_afternoon_ddos.rename(columns={column:column.strip()},inplace=True)
    
for column in friday_afternoon_portscan:
    friday_afternoon_portscan.rename(columns={column:column.strip()},inplace=True)
```

```python
monday.columns.to_list()
```

```python
# adding the sorce day column

monday['Source_day'] = 'monday_df'
tuesday['Source_day'] = 'tuesday_df'
wednesday['Source_day'] = 'wednesday_df'
thursday_morning['Source_day'] = 'thu_morning_df'
thursday_afternoon['Source_day'] = 'thu_afternoon_df'
friday_morning['Source_day'] = 'fri_morning_df'
friday_afternoon_ddos['Source_day'] = 'fri_afternoon_df'
friday_afternoon_portscan['Source_day'] = 'fri_afternoon_df'
```

```python
# Combining all the datasets so that we get one concrete data set
```

```python
full_df = pd.concat([
    monday,
    tuesday,
    wednesday,
    thursday_morning,
    thursday_afternoon,
    friday_morning,
    friday_afternoon_ddos,
    friday_afternoon_portscan
], ignore_index=True)

print("Final combined dataset shape:", full_df.shape)
```

```python
# Calculate Duplicate,null,missing Values

print(f'''
Duplicate values: {full_df.duplicated().sum()}\n
Null values: {full_df.isna().sum().sum()}\n
Missing values: {(full_df.isnull().sum().sum()/full_df.size)*100}%
''')
```

```python
# 1) Handling Duplicate Values

full_df.drop_duplicates(inplace=True)
```

```python
print(f'Null values percentage: {(full_df.isna().sum().sum()/full_df.size)*100}%\n')
```

```python
inf_counts = full_df.isin([np.inf, -np.inf]).sum()
print(inf_counts[inf_counts > 0])

total_inf = full_df.isin([np.inf, -np.inf]).sum().sum()
print(f"\nTotal infinite values: {total_inf}")
```

```python
# before dropping the isna values lets check if they cross the infinite values
bad_rows_mask = full_df.isna().any(axis=1) | full_df.isin([np.inf, -np.inf]).any(axis=1)
print(f"Total affected rows: {bad_rows_mask.sum()}")
print(f"Percentage of dataset: {bad_rows_mask.mean()*100:.4f}%")
```

```python
full_df = full_df[~bad_rows_mask].reset_index(drop=True)

print(full_df.shape)
print(f"Remaining nulls: {full_df.isna().sum().sum()}")
print(f"Remaining infinities: {full_df.isin([np.inf, -np.inf]).sum().sum()}")
```

```python
print(full_df['Label'].unique())
```

```python
full_df['attack_binary'] = (full_df['Label'] != 'BENIGN').astype(int)

print(full_df['attack_binary'].value_counts())
```

```python
# Check current encoding artifact
print(full_df['Label'].unique())

print()
# Fix - safe replace regardless of what the byte sequence actually is
full_df['Label'] = full_df['Label'].str.replace('�', '–', regex=False)
print(full_df['Label'].unique())
```

```python
print(full_df['Label'].value_counts())
```

```python
# Calculate the value counts
counts = full_df['attack_binary'].value_counts().sort_index()  
# Ensures 0 appears before 1

# Color mapping: 0 = Safe, 1 = Not Safe 
bar_colors = ['#51BD62', '#D61818']   

# Plot the counts
ax = counts.plot.bar(color=bar_colors, figsize=(8,5))

# Add data labels
ax.bar_label(ax.containers[0])

# Labels
plt.xlabel('Label\n0 = Safe    1 = Not Safe')
plt.ylabel('Count')
plt.title('Bar Chart of Safe vs Not Safe Samples (Training Data)')
plt.xticks(rotation=0)
plt.show()
```
